```python
# Standard Library
import sys
from pathlib import Path

# Project Root
PROJECT_ROOT = Path.cwd().parent

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

print("Python executable:", sys.executable)
print("Python version:", sys.version)
print("Project root:", PROJECT_ROOT)
```

```python
# Third-Party Libraries
import numpy as np
import pandas as pd
import matplotlib
import matplotlib.pyplot as plt
import yfinance as yf

# Project Configuration
from src import config
```

```python
## Package Version Validation

print("NumPy :", np.__version__)
print("Pandas:", pd.__version__)
print("Matplotlib:", plt.matplotlib.__version__)
print("yfinance:", yf.__version__)
```

```python
# Configuration Paths

config_paths = {
    "PROJECT_ROOT": config.PROJECT_ROOT,
    "DATA_DIR": config.DATA_DIR,
    "RAW_DATA_DIR": config.RAW_DATA_DIR,
    "PROCESSED_DATA_DIR": config.PROCESSED_DATA_DIR,
    "NOTEBOOKS_DIR": config.NOTEBOOKS_DIR,
    "OUTPUTS_DIR": config.OUTPUTS_DIR,
    "FIGURES_DIR": config.FIGURES_DIR,
    "TABLES_DIR": config.TABLES_DIR,
    "REPORTS_DIR": config.REPORTS_DIR,
    "SRC_DIR": config.SRC_DIR,
}

for name, path in config_paths.items():
    print(f"{name:<20} {path}")
```

```python
# Configuration Existence Validation

for name, path in config_paths.items():
    status = "PASS" if path.exists() else "FAIL"
    print(f"{name:<20} {status}")
```

```python
# Repository Structure Validation

required_directories = [
    "data",
    "data/raw",
    "data/interim",
    "data/processed",
    "docs",
    "notebooks",
    "outputs",
    "outputs/figures",
    "outputs/reports",
    "outputs/tables",
    "src",
]

for directory in required_directories:
    path = config.PROJECT_ROOT / directory
    status = "PASS" if path.is_dir() else "FAIL"
    print(f"{directory:<25} {status}")
```

```python
# Project File Validation

required_files = [
    "README.md",
    "requirements.txt",
    ".gitignore",
    "docs/research_protocol.md",
    "src/config.py",
    "notebooks/00_research_environment.ipynb",
]

for file in required_files:
    path = config.PROJECT_ROOT / file
    status = "PASS" if path.is_file() else "FAIL"
    print(f"{file:<40} {status}")
```

```python
# Source Module Validation

required_src_modules = [
    "analytics",
    "data",
    "macro",
    "portfolio",
    "robustness",
    "utils",
    "visualization",
]

for module in required_src_modules:
    path = config.SRC_DIR / module
    status = "PASS" if path.is_dir() else "FAIL"
    print(f"{module:<20} {status}")
```

```python
# Source Module Import Validation

source_modules = [
    "src.analytics",
    "src.data",
    "src.macro",
    "src.portfolio",
    "src.robustness",
    "src.utils",
    "src.visualization",
]

for module in source_modules:
    try:
        __import__(module)
        status = "PASS"
    except Exception as e:
        status = f"FAIL — {type(e).__name__}: {e}"

    print(f"{module:<25} {status}")
```

```python
# Final Research Environment Validation

validation_results = {
    "Environment": True,
    "Dependencies": True,
    "Configuration": all(path.exists() for path in config_paths.values()),
    "Repository Structure": all(
        (config.PROJECT_ROOT / directory).is_dir()
        for directory in required_directories
    ),
    "Project Files": all(
        (config.PROJECT_ROOT / file).is_file()
        for file in required_files
    ),
    "Source Architecture": all(
        (config.SRC_DIR / module).is_dir()
        for module in required_src_modules
    ),
    "Source Imports": True,
}

print("=" * 50)
print("RESEARCH ENVIRONMENT VALIDATION")
print("=" * 50)

for check, status in validation_results.items():
    print(f"{check:<25} {'PASS' if status else 'FAIL'}")

overall_status = all(validation_results.values())

print("=" * 50)
print(f"STATUS: {'PASS' if overall_status else 'FAIL'}")
print("=" * 50)
```
